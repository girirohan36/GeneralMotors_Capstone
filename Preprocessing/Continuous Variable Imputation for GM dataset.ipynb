```python
import pandas as pd
```

```python
# Load the summary statistics and actual dataset
remaining_data_after_step2 = "remaining_data_after_step2.csv"  
filtered_actual_dataset = "filtered_actual_dataset.csv"  

summary_stats_df = pd.read_csv(remaining_data_after_step2)
filtered_actual_dataset = pd.read_csv(filtered_actual_dataset)
```

# Datasets Used:
remaining_data_after_step2.csv: Contains the 99 remaining features along with their summary statistics (e.g., Mean, Median, Pct_25, Pct_75).
filtered_actual_dataset.csv: The filtered actual dataset with tens of thousands of rows and only the 99 remaining features.
Goal: Impute missing values in numerical columns based on their distributions.

```python
# Function to decide imputation method based on summary statistics
def decide_imputation_method(summary_stats, feature_name):
    row = summary_stats[summary_stats['Feature'] == feature_name]
    if not row.empty:
        mean = row['Mean'].values[0]
        median = row['Median'].values[0]
        # Check symmetry using mean and median
        if abs(mean - median) / max(abs(mean), abs(median)) < 0.1:  # Symmetry threshold
            return "mean"
        else:
            return "median"
    return None
```

# Imputation Method:

The choice of imputation method for each numerical feature was determined using the summary statistics:
Mean Imputation:
Applied to symmetric distributions where the Mean is close to the Median.
Symmetry was assessed using:
Symmetry Check: 

Median Imputation:
Applied to skewed distributions where the Mean differs significantly from the Median.

```python
# Impute missing values in numerical features only
for feature in filtered_actual_dataset.columns:
    if pd.api.types.is_numeric_dtype(filtered_actual_dataset[feature]):  # Check if the column is numeric
        method = decide_imputation_method(summary_stats_df, feature)
        if method == "mean":
            filtered_actual_dataset[feature].fillna(filtered_actual_dataset[feature].mean(), inplace=True)
        elif method == "median":
            filtered_actual_dataset[feature].fillna(filtered_actual_dataset[feature].median(), inplace=True)
```

```python
# Save the imputed dataset (optional)
output_file_path = "imputed_actual_dataset_numerical.csv"
filtered_actual_dataset.to_csv(output_file_path, index=False)

print(f"Imputed dataset with numerical features saved to: {output_file_path}")
```
