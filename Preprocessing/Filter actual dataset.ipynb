```python
import pandas as pd
```

```python
# Load the remaining features after Step 2
remaining_features_path = "remaining_data_after_step2.csv"  # Update with correct path
remaining_features = pd.read_csv(remaining_features_path)

# Extract feature names
filtered_feature_names = remaining_features['Feature'].tolist()  # Ensure the column name is 'Feature'
```

```python
# Load the actual dataset
actual_dataset_path = "EV VIO.csv"  # Update with correct path
actual_dataset = pd.read_csv(actual_dataset_path)
```

```python
# Subset the actual dataset to keep only the filtered feature names
filtered_actual_dataset = actual_dataset[filtered_feature_names]
```

```python
# Save the filtered actual dataset
output_file_path = "filtered_actual_dataset.csv"
filtered_actual_dataset.to_csv(output_file_path, index=False)

print(f"Filtered actual dataset saved to: {output_file_path}")
```
